**Cuaderno editable para Google Colab.** Explora qué años se parecen a mayo–septiembre de 2026 usando cuatro índices NOAA.

Sube el ZIP correspondiente a este módulo cuando la primera celda lo solicite. Los gráficos se generan a partir de los CSV originales.

```python
from google.colab import files
from pathlib import Path
import zipfile, io
print('Selecciona el ZIP individual de este módulo:')
uploaded = files.upload()
zip_names = [n for n in uploaded if n.lower().endswith('.zip')]
if len(zip_names) != 1:
    raise ValueError('Sube exactamente un ZIP de este módulo.')
with zipfile.ZipFile(io.BytesIO(uploaded[zip_names[0]])) as z:
    z.extractall('/content/entrega')
base = Path('/content/entrega') / 'ML1_Historia_El_Nino'
if not (base / 'app.py').exists():
    raise FileNotFoundError('No se encontró app.py en el ZIP esperado.')
print('Archivos encontrados:', base)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
plt.rcParams.update({"figure.figsize":(10,5),"font.size":11})
```

```python
df = pd.read_csv(base/'datos'/'enso_dataset_maestro.csv')
ranking = pd.read_csv(base/'datos'/'01_ranking_similitud.csv')
display(ranking.head(10))
fig, ax = plt.subplots()
for year, color in [(2026,'#E58A36'),(1997,'#087E8B'),(2015,'#57748A'),(2023,'#B94E48')]:
    q = df[(df.anio==year)&df.mes.between(5,9)].sort_values('mes')
    ax.plot(q.mes,q.anom_nino34,marker='o',linewidth=3 if year==2026 else 2,label=str(year),color=color)
ax.axhline(0,color='gray',linewidth=1)
ax.set_xticks(range(5,10),['Mayo','Junio','Julio','Agosto','Septiembre'])
ax.set_ylabel('Anomalía Niño 3.4 (°C)')
ax.set_title('ML 1 | Comparación histórica NOAA')
ax.legend(); ax.grid(alpha=.2); plt.show()
```
