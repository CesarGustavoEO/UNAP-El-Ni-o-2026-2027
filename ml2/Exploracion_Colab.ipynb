```python
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display
plt.rcParams.update({"figure.figsize":(10,5),"font.size":11})
```

```python
comparacion = pd.read_csv(base/'datos'/'02_comparacion_nacional.csv')
pred = pd.read_csv(base/'datos'/'01_predicciones_retrospectivas.csv')
display(comparacion.sort_values('aporte_ENSO_pct',ascending=False).head(10))
orden = comparacion.sort_values('aporte_ENSO_pct')
fig, ax = plt.subplots(figsize=(10,8))
colors = ['#087E8B' if v>0 else '#B94E48' for v in orden.aporte_ENSO_pct]
ax.barh(orden.departamento,orden.aporte_ENSO_pct,color=colors)
ax.axvline(0,color='black',linewidth=1)
ax.set_xlabel('Reducción del error al añadir ENSO (%)')
ax.set_title('ML 2 | ¿Dónde ayudó ENSO al modelo?')
plt.tight_layout(); plt.show()
# Vista ciudadana: lluvia observada frente a climatología mensual histórica
clima = pd.read_csv(base/'datos'/'02_dataset_nacional_noaa_power.csv')
region = 'Puno'
q=clima[(clima.departamento==region)&(clima.anio==2023)].sort_values('mes')
ref=clima[(clima.departamento==region)&clima.anio.between(2001,2018)].groupby('mes').lluvia_mm_mes.mean()
fig,ax=plt.subplots()
ax.bar(q.mes,q.lluvia_mm_mes,color='#087E8B',label='Lluvia observada en el punto')
ax.plot(ref.index,ref.values,color='#E58A36',marker='o',label='Promedio 2001–2018')
ax.set_xticks(range(1,13)); ax.set_xlabel('Mes'); ax.set_ylabel('Precipitación (mm/mes)')
ax.set_title('Puno (punto de referencia) | 2023')
ax.legend(); plt.show()
```
